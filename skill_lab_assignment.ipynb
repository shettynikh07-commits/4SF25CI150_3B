```python
import pandas as pd
import numpy as np

data = np.random.randint(1, 10, size=(4, 3))

df = pd.DataFrame(data, columns=['A', 'B', 'C'])

print("DataFrame:")
print(df)

print("Row-wise Sum:")
print(df.sum(axis=1))

print("Column-wise Sum:")
print(df.sum(axis=0))
```

```python
import pandas as pd
import numpy as np

data = np.random.randint(1, 10, size=(5, 3))

df = pd.DataFrame(data, columns=['A', 'B', 'C'])

df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan

print("Original DataFrame:")
print(df)

df = df.fillna(df.mean())

print("DataFrame after filling NaN:")
print(df)
```

```python
import pandas as pd
import numpy as np

data = np.random.randint(1, 10, size=(6, 4))

df = pd.DataFrame(data, columns=['A', 'B', 'C', 'D'])

df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan

print("Original DataFrame:")
print(df)

df = df.dropna()

print("After dropping NaN rows:")
print(df)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(1, 100, 10)
})

print("Original DataFrame:")
print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("Grouped Data:")
print(result)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Pencil', 'Bag', 'Eraser'],
    'Category': ['Stationery', 'Books', 'Stationery', 'Bags', 'Stationery'],
    'Sales': np.random.randint(100, 1000, 5)
})

print("Original DataFrame:")
print(df)

result = df.groupby('Category')['Sales'].sum()

print("Total Sales:")
print(result)
```

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Amit', 'Ravi', 'Sita']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [80, 90, 85]
})

result = pd.merge(df1, df2, on='ID')

print(result)
```

```python
import pandas as pd

index = pd.MultiIndex.from_tuples(
    [('A', 1), ('A', 2), ('B', 1), ('B', 2)],
    names=['Category', 'Number']
)

df = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)

print("Original DataFrame:")
print(df)

print("Indexing:")
print(df.loc['A'])

print("Slicing:")
print(df.loc[('A', 1):('B', 1)])
```

```python
import pandas as pd
import numpy as np

index = pd.MultiIndex.from_product(
    [['A', 'B'], ['X', 'Y']],
    names=['Category', 'SubCategory']
)

df = pd.DataFrame({
    'Value': np.random.randint(1, 100, 4)
}, index=index)

print("Original DataFrame:")
print(df)

print("Sum:")
print(df.groupby(level=['Category', 'SubCategory'])['Value'].sum())
```

```python
import pandas as pd
import numpy as np

data = np.random.randint(1, 10, size=(5, 3))

df = pd.DataFrame(data, columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

result = df.apply(lambda x: x * 2)

print("After Doubling:")
print(result)
```

```python
import pandas as pd
import numpy as np

data = np.random.randint(1, 10, size=(6, 3))

df = pd.DataFrame(data, columns=['A', 'B', 'C'])

df['Total'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)

print(df)
```

```python
import pandas as pd

data = pd.Series(['apple', 'banana', 'mango', 'orange', 'grapes'])

result = data.str.upper()

print("Original Series:")
print(data)

print("Uppercase Series:")
print(result)
```
